1. One aggregate summary row was identified through
   `CUSTOMER_CODE = 'Total'`.

   than individual transaction records.

   based on the dataset's stated assumption.

### Data Preparation Decisions

- Preserve the original raw dataset.
- Exclude the aggregate summary row from transaction-level
  analysis.
- Standardize column names in a working copy.
- Convert numeric fields into appropriate numeric data types.
- Reconcile transaction totals against the source summary
  before proceeding with further cleaning.

```python

from decimal import Decimal, ROUND_HALF_UP

def round_half_up(value, decimal_places=1):
    if pd.isna(value):
        return np.nan

    decimal_value = Decimal(str(value))
    rounding_unit = Decimal("1").scaleb(-decimal_places)

    return float(
        decimal_value.quantize(
            rounding_unit,
            rounding=ROUND_HALF_UP
        )
    )

# Terapkan pembulatan Half-Up
df_transactions_raw["Half Up Implied Price"] = (
    df_transactions_raw["Implied Selling Price"]
    .apply(round_half_up)
)

# Bandingkan dengan Selling price
df_transactions_raw["Half Up Match"] = np.isclose(
    df_transactions_raw["Half Up Implied Price"],
    df_transactions_raw["Selling price"],
    atol=0.001,
    rtol=0
)

# Ringkasan hasil
half_up_test = pd.DataFrame({
    "Rows Tested": [len(df_transactions_raw)],
    "Matching Rows": [
        df_transactions_raw["Half Up Match"].sum()
    ],
    "Non-Matching Rows": [
        (~df_transactions_raw["Half Up Match"]).sum()
    ]
}, index=["Half-Up Rounding Test"])

half_up_test
```

```python

# Simulasi pembulatan Excel ROUND untuk angka positif
def excel_round(value, decimal_places=1):
    if pd.isna(value):
        return np.nan

    multiplier = Decimal("10") ** decimal_places
    decimal_value = Decimal(str(value))

    result = (
        (decimal_value * multiplier)
        .quantize(Decimal("1"), rounding=ROUND_HALF_UP)
        / multiplier
    )

    return float(result)

df_transactions_raw["Excel Rounded Price"] = (
    df_transactions_raw["Implied Selling Price"]
    .apply(excel_round)
)

df_transactions_raw["Excel Rounding Match"] = np.isclose(
    df_transactions_raw["Excel Rounded Price"],
    df_transactions_raw["Selling price"],
    atol=0.001,
    rtol=0
)

excel_rounding_test = pd.DataFrame({
    "Rows Tested": [len(df_transactions_raw)],
    "Matching Rows": [
        df_transactions_raw["Excel Rounding Match"].sum()
    ],
    "Non-Matching Rows": [
        (~df_transactions_raw["Excel Rounding Match"]).sum()
    ]
}, index=["Excel Rounding Test"])

excel_rounding_test
```

```python

def calculate_decimal_price(row):
    sales = Decimal(str(row["SALES"]))
    quantity = Decimal(str(int(row["QUANTITY_ORDERED"])))

    implied_price = sales / quantity

    return float(
        implied_price.quantize(
            Decimal("0.1"),
            rounding=ROUND_HALF_UP
        )
    )

df_transactions_raw["Decimal Rounded Price"] = (
    df_transactions_raw.apply(calculate_decimal_price, axis=1)
)

df_transactions_raw["Decimal Rounding Match"] = np.isclose(
    df_transactions_raw["Decimal Rounded Price"],
    df_transactions_raw["Selling price"],
    atol=0.001,
    rtol=0
)

decimal_rounding_test = pd.DataFrame({
    "Rows Tested": [len(df_transactions_raw)],
    "Matching Rows": [
        df_transactions_raw["Decimal Rounding Match"].sum()
    ],
    "Non-Matching Rows": [
        (~df_transactions_raw["Decimal Rounding Match"]).sum()
    ]
}, index=["Decimal Half-Up Test"])

decimal_rounding_test
```

```python

# Buat status berdasarkan nilai profit/loss
def classify_profit_status(value):
    if pd.isna(value):
        return "Missing"

    if value > 0:
        return "Profit"
    elif value < 0:
        return "Loss"
    else:
        return "Break-even"

df_transactions_raw["Calculated Status"] = (
    df_transactions_raw["Total profit / loss"]
    .apply(classify_profit_status)
)

# Bandingkan dengan status asli
df_transactions_raw["Status Clean"] = (
    df_transactions_raw["Status"]
    .astype(str)
    .str.strip()
    .str.title()
)

df_transactions_raw["Status Match"] = (
    df_transactions_raw["Status Clean"]
    == df_transactions_raw["Calculated Status"]
)

# Ringkasan distribusi status
status_validation = pd.DataFrame({
    "Source Status": df_transactions_raw["Status Clean"].value_counts(),
    "Calculated Status": df_transactions_raw["Calculated Status"].value_counts()
}).fillna(0).astype(int)

print("Distribusi Status:")
display(status_validation)

# Ringkasan validasi
print("\nStatus Validation:")
display(
    pd.DataFrame({
        "Rows Tested": [len(df_transactions_raw)],
        "Matching Rows": [df_transactions_raw["Status Match"].sum()],
        "Non-Matching Rows": [
            (~df_transactions_raw["Status Match"]).sum()
        ]
    }, index=["Status Consistency Test"])
)
```
